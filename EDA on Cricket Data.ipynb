### Data Cleaning and Preprocessing

```python
# Missing values
df.isnull().sum()
```

```python
# Visualize missing data
sns.heatmap(df.isnull(), cmap='viridis')
```

```python
# Checking for duplicates
df.duplicated().sum()
```

```python
# Removing duplicates
df = df.drop_duplicates()
```

```python
df = df.replace(['–', '-', 'NA', 'na', '', '*'], np.nan)
```

```python
df
```

```python
# Removing * from Highest Score
df['Highest_Score'] = df['Highest_Score'].str.replace(r'\D$', '', regex=True)
```

```python
# Identifying numeric column
non_numeric = ['Player_Name', 'Best_Bowling_Match']
numerical_cols = [col for col in df.columns if col not in non_numeric]
```

```python
numerical_cols
```

```python
# converting to numerical
df[numerical_cols] = df[numerical_cols].apply(pd.to_numeric, errors='coerce')
```

```python
column_types = df[numerical_cols].dtypes
column_types
```

```python
# Normalizing and standardizing
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
df_scaled = df.copy()
df_scaled[numerical_cols] = scaler.fit_transform(df[numerical_cols])
```

###  Exploratory Data Analysis (EDA)

```python
# Distribution of Batting Average
plt.figure(figsize=(8, 5))
sns.histplot(df['Batting_Average'].dropna(), bins=30, kde=True)
plt.title("Distribution of Batting Average")
plt.xlabel("Batting Average")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

```python
# Player with highest Batting Strike Rate
highest_strike_rate = df.loc[df['Batting_Strike_Rate'].idxmax(), ['Player_Name', 'Batting_Strike_Rate']]
print("Player with Highest Batting Strike Rate:\n", highest_strike_rate)
```

```python
# Trend in Runs Scored for each player over years (here if we creating a all players trends graph then it mixes with all)
top_players = df.groupby("Player_Name")["Runs_Scored"].sum().nlargest(5).index.tolist()
df_trend = df[df["Player_Name"].isin(top_players)]

plt.figure(figsize=(10, 6))
sns.lineplot(data=df_trend, x="Year", y="Runs_Scored", hue="Player_Name", marker="o")
plt.title("Runs Scored Over Years (Top 5 Players)")
plt.xlabel("Year")
plt.ylabel("Runs Scored")
plt.tight_layout()
plt.show()
```

```python
# Top 3 players with most Centuries
top_centuries = df.groupby("Player_Name")["Centuries"].sum().nlargest(3).reset_index()
print("Top 3 Players with Most Centuries:\n", top_centuries)
```

```python
# Player with the highest Bowling Economy Rate
highest_economy = df.loc[df['Economy_Rate'].idxmax(), ['Player_Name', 'Economy_Rate']]
print("Player with Highest Bowling Economy Rate:\n", highest_economy)
```

```python
# Boundary Percentage
df["Boundary_Runs"] = (df["Fours"] * 4) + (df["Sixes"] * 6)

df["Boundary_Percentage"] = (df["Boundary_Runs"] / df["Runs_Scored"]) * 100
```

```python
df["Boundary_Percentage"].head()
```

```python
# Batting Impact Score
df["Batting_Impact_Score"] = (
    df["Runs_Scored"] * 0.5 +
    df["Batting_Strike_Rate"] * 0.3 +
    df["Batting_Average"] * 0.2
)
```

```python
df.head()
```

```python
# All-round performance considering both batting and bowling stats.
from sklearn.preprocessing import MinMaxScaler

features = ["Runs_Scored", "Batting_Strike_Rate", "Batting_Average", 
            "Wickets_Taken", "Economy_Rate", "Bowling_Strike_Rate"]

scaler = MinMaxScaler()
df_scaled_feats = pd.DataFrame(scaler.fit_transform(df[features]), columns=features)

df["All_Round_Score"] = (
    df_scaled_feats["Runs_Scored"] * 0.3 +
    df_scaled_feats["Batting_Strike_Rate"] * 0.2 +
    df_scaled_feats["Batting_Average"] * 0.2 +
    df_scaled_feats["Wickets_Taken"] * 0.2 +
    (1 - df_scaled_feats["Economy_Rate"]) * 0.05 +
    (1 - df_scaled_feats["Bowling_Strike_Rate"]) * 0.05 
)
```

```python
df.head()
```

```python
# Categorize players into Aggressive, Balanced, and Defensive based on their strike rate.
def categorize_strike_rate(sr):
    if sr >= 140:
        return "Aggressive"
    elif sr >= 120:
        return "Balanced"
    else:
        return "Defensive"

df["Batting_Style"] = df["Batting_Strike_Rate"].apply(categorize_strike_rate)
```

```python
df.head()
```

```python
# Compute the year-over-year growth in performance metrics for each player.
df = df.sort_values(by=["Player_Name", "Year"])

df["YoY_Runs_Growth"] = df.groupby("Player_Name")["Runs_Scored"].pct_change() * 100

df["YoY_StrikeRate_Growth"] = df.groupby("Player_Name")["Batting_Strike_Rate"].pct_change() * 100
df["YoY_Wickets_Growth"] = df.groupby("Player_Name")["Wickets_Taken"].pct_change() * 100
```

```python
# Create a bar chart to show the Runs Scored vs. Balls Faced for all players.
runs_vs_balls = df.groupby("Player_Name")[["Runs_Scored", "Balls_Faced"]].sum().sort_values(by="Runs_Scored", ascending=False).head(20)

runs_vs_balls.plot(kind='bar', figsize=(12, 6))
plt.title("Runs Scored vs Balls Faced (Top 20 Players)")
plt.ylabel("Total")
plt.xlabel("Player Name")
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
# Generate a heatmap showing the correlation between different numerical features
plt.figure(figsize=(14, 10))
sns.heatmap(df.corr(numeric_only=True), annot=True, fmt=".2f", cmap="coolwarm")
plt.title("Correlation Heatmap of Numerical Features")
plt.tight_layout()
plt.show()
```

```python
# Plot a trend line showing the performance of a single player over multiple years.
player_name = "Virat Kohli" 
player_data = df[df["Player_Name"] == player_name].sort_values("Year")

plt.figure(figsize=(8, 5))
sns.lineplot(data=player_data, x="Year", y="Runs_Scored", marker='o')
plt.title(f"Runs Scored Over Years - {player_name}")
plt.xlabel("Year")
plt.ylabel("Runs Scored")
plt.tight_layout()
plt.show()
```

```python
# Create a pie chart for the distribution of centuries and half-centuries across all players.
total_centuries = df["Centuries"].sum()
total_half_centuries = df["Half_Centuries"].sum()

plt.figure(figsize=(6, 6))
plt.pie(
    [total_centuries, total_half_centuries],
    labels=["Centuries", "Half-Centuries"],
    autopct="%1.1f%%",
    startangle=90,
    colors=["#1f77b4", "#ff7f0e"]
)
plt.title("Distribution of Centuries vs Half-Centuries")
plt.tight_layout()
plt.show()
```

```python
# Develop a visualization to compare Bowling Average vs. Bowling Strike Rate.
plt.figure(figsize=(8, 6))
sns.scatterplot(data=df, x="Bowling_Strike_Rate", y="Bowling_Average", hue="Player_Name", legend=False)
plt.title("Bowling Average vs Bowling Strike Rate")
plt.xlabel("Bowling Strike Rate")
plt.ylabel("Bowling Average")
plt.tight_layout()
plt.show()
```

```python
# top 5 players with the highest Batting Average.
top_5_batting_avg = df[['Player_Name', 'Batting_Average']] \
    .sort_values(by='Batting_Average', ascending=False) \
    .dropna(subset=['Batting_Average']) \
    .head(5)

print("Top 5 Players by Batting Average:\n", top_5_batting_avg)
```

```python
# which player has played the most matches.
df['Total_Matches'] = df['Matches_Batted'] + df['Matches_Bowled']

most_matches = df.groupby('Player_Name')['Total_Matches'].sum() \
    .sort_values(ascending=False).head(1)

print("Player with Most Total Matches:\n", most_matches)
```

```python
# Players Who Scored More Than 500 Runs in a Season
high_scorers = df[df['Runs_Scored'] > 500][['Year', 'Player_Name', 'Runs_Scored']] \
    .sort_values(by='Runs_Scored', ascending=False)

print("Players with More Than 500 Runs in a Season:\n", high_scorers)
```

```python
# Players with At Least One Five-Wicket Haul
five_hauls = df[df['Five_Wicket_Hauls'] >= 1]['Player_Name'].unique()

print("Players with ≥ 1 Five-Wicket Haul:\n", five_hauls)
```

```python
# Total number of boundaries hit by each player.
df['Total_Boundary_Runs'] = (df['Fours'] * 4) + (df['Sixes'] * 6)

boundaries_per_player = df.groupby('Player_Name')['Total_Boundary_Runs'].sum() \
    .sort_values(ascending=False).reset_index()

print("Total Boundary Runs by Each Player:\n", boundaries_per_player)
```
